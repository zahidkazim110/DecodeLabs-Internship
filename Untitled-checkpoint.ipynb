```python
from tkinter import *
from tkinter import scrolledtext
from datetime import datetime

# ==========================
# SmartBot Knowledge Base
# ==========================

responses = {
    "hello": "👋 Hello! Welcome to SmartBot.",
    "hi": "😊 Hi there! How can I help you today?",
    "hey": "🙌 Hey! Hope you're having a great day.",
    "good morning": "☀️ Good morning!",
    "good afternoon": "🌤️ Good afternoon!",
    "good evening": "🌙 Good evening!",
    "how are you": "😄 I'm doing great! Thanks for asking.",
    "what is your name": "🤖 My name is SmartBot AI.",
    "who are you": "🤖 I am a Rule-Based AI Chatbot.",
    "who created you": "👨‍💻 Created for DecodeLabs AI Internship.",
    "what can you do": "🚀 I can chat, tell date & time, perform calculations, motivate you and tell jokes.",
    "tell me a joke": "😂 Why do programmers prefer dark mode? Because light attracts bugs!",
    "tell me another joke": "🤣 Why was the computer cold? Because it left its Windows open!",
    "motivate me": "💪 Success is the sum of small efforts repeated every day.",
    "thank you": "😊 You're welcome!",
    "thanks": "😎 Happy to help!",
    "python": "🐍 Python is one of the most popular programming languages.",
    "ai": "🧠 Artificial Intelligence enables machines to mimic human intelligence.",
    "machine learning": "📊 Machine Learning is a branch of AI that learns from data.",
    "favorite language": "❤️ Python is definitely my favorite language!"
}

# ==========================
# Chatbot Logic
# ==========================

def chatbot_response(message):

    msg = message.lower().strip()

    if msg == "date":
        return f"📅 Today's Date: {datetime.now().strftime('%d-%m-%Y')}"

    if msg == "time":
        return f"⏰ Current Time: {datetime.now().strftime('%H:%M:%S')}"

    if msg.startswith("add"):
        try:
            _, a, b = msg.split()
            return f"➕ Result = {float(a)+float(b)}"
        except:
            return "❌ Usage: add 10 20"

    if msg.startswith("subtract"):
        try:
            _, a, b = msg.split()
            return f"➖ Result = {float(a)-float(b)}"
        except:
            return "❌ Usage: subtract 20 10"

    if msg.startswith("multiply"):
        try:
            _, a, b = msg.split()
            return f"✖️ Result = {float(a)*float(b)}"
        except:
            return "❌ Usage: multiply 5 6"

    if msg.startswith("divide"):
        try:
            _, a, b = msg.split()

            if float(b) == 0:
                return "❌ Cannot divide by zero"

            return f"➗ Result = {float(a)/float(b)}"
        except:
            return "❌ Usage: divide 20 4"

    return responses.get(
        msg,
        "🤖 Sorry, I don't understand that command."
    )

# ==========================
# Send Message Function
# ==========================

def send_message():

    message = user_entry.get()

    if message.strip() == "":
        return

    chat_area.config(state=NORMAL)

    chat_area.insert(END, f"\n🧑 You: {message}\n")

    response = chatbot_response(message)

    chat_area.insert(END, f"🤖 SmartBot: {response}\n")

    chat_area.config(state=DISABLED)

    user_entry.delete(0, END)

    chat_area.yview(END)

# ==========================
# Enter Key Support
# ==========================

def enter_pressed(event):
    send_message()

# ==========================
# Main Window
# ==========================

root = Tk()

root.title("SmartBot AI Assistant")
root.geometry("900x650")
root.config(bg="#0f172a")

# ==========================
# Header
# ==========================

title_label = Label(
    root,
    text="🤖 SmartBot AI Assistant",
    font=("Segoe UI", 22, "bold"),
    bg="#0f172a",
    fg="white"
)

title_label.pack(pady=10)

subtitle_label = Label(
    root,
    text="DecodeLabs Artificial Intelligence Internship Project 1",
    font=("Segoe UI", 11),
    bg="#0f172a",
    fg="#94a3b8"
)

subtitle_label.pack()

# ==========================
# Chat Area
# ==========================

chat_area = scrolledtext.ScrolledText(
    root,
    wrap=WORD,
    font=("Segoe UI", 11),
    bg="white",
    fg="black"
)

chat_area.pack(
    padx=20,
    pady=20,
    fill=BOTH,
    expand=True
)

chat_area.insert(
    END,
    "🤖 SmartBot: Hello! Welcome to SmartBot AI.\n"
    "Type a message below to start chatting.\n"
)

chat_area.config(state=DISABLED)

# ==========================
# Bottom Frame
# ==========================

bottom_frame = Frame(root, bg="#0f172a")
bottom_frame.pack(fill=X, padx=20, pady=10)

user_entry = Entry(
    bottom_frame,
    font=("Segoe UI", 12)
)

user_entry.pack(
    side=LEFT,
    fill=X,
    expand=True,
    ipady=8
)

user_entry.bind("<Return>", enter_pressed)

send_button = Button(
    bottom_frame,
    text="Send",
    command=send_message,
    font=("Segoe UI", 11, "bold"),
    bg="#2563eb",
    fg="white",
    padx=20
)

send_button.pack(side=RIGHT, padx=10)

# ==========================
# Footer
# ==========================

footer = Label(
    root,
    text="Developed for DecodeLabs AI Internship 2026",
    font=("Segoe UI", 9),
    bg="#0f172a",
    fg="#94a3b8"
)

footer.pack(pady=5)

root.mainloop()
```
